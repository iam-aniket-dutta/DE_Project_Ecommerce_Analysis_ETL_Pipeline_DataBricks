# Unity Catalog Lakehouse Setup: Ecommerce
This notebook initializes the catalog, medallion architecture schemas (`bronze`, `silver`, `gold`), and managed volumes for raw data ingestion.

```python
%%sql
-- 1. Create Catalog
CREATE CATALOG IF NOT EXISTS ecommerce
COMMENT 'Lakehouse catalog for Ecommerce Analytics & BI';
```

```python
%%sql
-- 2. Select Active Catalog
USE CATALOG ecommerce;
```

```python
%%sql
-- 3. Create Medallion Schemas
CREATE SCHEMA IF NOT EXISTS ecommerce.bronze
COMMENT 'Raw ingested data from source files and landing volumes';

CREATE SCHEMA IF NOT EXISTS ecommerce.silver
COMMENT 'Cleaned, conformed, and validated business entities';

CREATE SCHEMA IF NOT EXISTS ecommerce.gold
COMMENT 'Aggregated, star-schema dimensional models and BI-ready tables';

-- Backward compatibility schema for legacy pipelines referencing 'broze'
CREATE SCHEMA IF NOT EXISTS ecommerce.broze
COMMENT 'Backward-compatibility alias schema for legacy code referencing broze';
```

```python
%%sql
-- 4. Create Unity Catalog Managed Volumes for Raw Ingestion
CREATE VOLUME IF NOT EXISTS ecommerce.bronze.raw
COMMENT 'Managed volume storing raw CSV landing files (brands, category, customers, date, products, order_items)';
```

```python
%%sql
-- 5. Verify Setup
SHOW SCHEMAS IN ecommerce;
SHOW VOLUMES IN ecommerce.bronze;
```
